```python
import numpy as np

# ==========================
# 读取数据
# ==========================
input_file = "Type_B_MULTISOURCE_1000_128_GT.npy"
output_file = "./tmp/TYPEB_MULTISOURCE_1000_128.npy"

data = np.load(input_file)

print("Original shape:", data.shape)

# ==========================
# 1. 统一维度
# ==========================
if data.ndim == 3:
    # (N,H,W) -> (N,1,H,W)
    data = data[:, np.newaxis, :, :]

elif data.ndim == 4:
    # 已经是 (N,C,H,W)
    pass

else:
    raise ValueError(
        f"Unsupported shape {data.shape}. "
        "Expected (N,H,W) or (N,1,H,W)."
    )

print("After reshape:", data.shape)

# 转成 float32
data = data.astype(np.float32)

# ==========================
# 2. 每张图总和归一化
# ==========================
# shape = (N,1,H,W)

img_sum = data.sum(axis=(2, 3), keepdims=True)

# 防止全零图
if np.any(img_sum == 0):
    raise ValueError(
        f"Found {np.sum(img_sum == 0)} image(s) with total flux = 0."
    )

data_norm = data / img_sum

# ==========================
# 检查结果
# ==========================
check_sum = data_norm.sum(axis=(2, 3))

print("Sum range:")
print("min =", check_sum.min())
print("max =", check_sum.max())
print("mean =", check_sum.mean())
# ==========================
# 3. 保存
# ==========================
np.save(output_file, data_norm)

print(f"Saved to: {output_file}")
print("Final shape:", data_norm.shape)
```
