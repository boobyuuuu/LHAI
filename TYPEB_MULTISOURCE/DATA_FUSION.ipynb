```python
import numpy as np

# ==========================================
# 需要融合的数据集
# ==========================================
FUSION_DATA = [
    "./tmp/TYPEA_COMPACT_1000_128_left.npy",
    "./tmp/TYPEA_COMPACT_1000_128_up.npy",
]

OUTPUT_FILE = "./tmp/TYPEA_COMPACT_1000_128_leftup.npy"

# ==========================================
# 读取并检查
# ==========================================
datasets = []

for file in FUSION_DATA:
    data = np.load(file)

    print(f"{file}: {data.shape}")

    datasets.append(data)

# ==========================================
# 检查除第0维外是否一致
# ==========================================
ref_shape = datasets[0].shape[1:]

for i, data in enumerate(datasets):
    if data.shape[1:] != ref_shape:
        raise ValueError(
            f"Shape mismatch:\n"
            f"{FUSION_DATA[0]} -> {datasets[0].shape}\n"
            f"{FUSION_DATA[i]} -> {data.shape}"
        )

# ==========================================
# 融合
# ==========================================
fusion_data = np.concatenate(datasets, axis=0)

print("\nFusion complete")
print("Output shape:", fusion_data.shape)

# ==========================================
# 保存
# ==========================================
np.save(OUTPUT_FILE, fusion_data)

print(f"Saved to: {OUTPUT_FILE}")
```
